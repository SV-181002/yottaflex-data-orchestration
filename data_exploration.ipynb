```python
%%writefile src/generate_data.py

"""
Synthetic E-commerce Data Generator

Purpose:
Generate reproducible customer and order datasets.
The generator supports normal records and deliberately
introduced data-quality problems for pipeline testing.
"""

# Import standard Python libraries.
import argparse
import random

from pathlib import Path
from datetime import datetime, timedelta, timezone

# Import external libraries.
import pandas as pd
from faker import Faker


def generate_data(
    customer_count=100,
    order_count=200,
    seed=42,
    inject_issues=True
):
    """
    Generate synthetic customers and orders.

    Parameters:
        customer_count: Number of customers to generate.
        order_count: Number of orders to generate.
        seed: Fixed random seed for reproducibility.
        inject_issues: Whether to introduce test anomalies.

    Returns:
        customers_df: Customer DataFrame.
        orders_df: Order DataFrame.
        expected_rejects_df: Separate expected rejection labels.
    """

    # Set fixed seeds so the generated data is reproducible.
    random.seed(seed)
    Faker.seed(seed)

    fake = Faker()
    rng = random.Random(seed)

    # Define the customer records.
    customers = []

    for i in range(1, customer_count + 1):

        customer_id = f"C{i:06d}"

        customers.append({
            "customer_id": customer_id,
            "customer_name": fake.name(),
            "email": f"{customer_id.lower()}@example.com",
            "city": fake.city(),
            "created_at": (
                datetime(2025, 1, 1, tzinfo=timezone.utc)
                + timedelta(days=rng.randint(0, 365))
            ).isoformat()
        })

    # Convert customer records into a DataFrame.
    customers_df = pd.DataFrame(customers)

    # Store the available customer IDs.
    valid_customer_ids = customers_df["customer_id"].tolist()

    # Define allowed order statuses.
    statuses = [
        "PLACED",
        "SHIPPED",
        "DELIVERED",
        "CANCELLED"
    ]

    # Generate order records.
    orders = []

    for i in range(1, order_count + 1):

        order_id = f"O{i:07d}"

        orders.append({
            "source_record_id": f"SRC{i:07d}",
            "order_id": order_id,
            "customer_id": rng.choice(valid_customer_ids),
            "order_timestamp": (
                datetime(2025, 1, 1, tzinfo=timezone.utc)
                + timedelta(days=rng.randint(0, 365))
            ).isoformat(),
            "quantity": rng.randint(1, 10),
            "unit_price": round(rng.uniform(10, 1000), 2),
            "status": rng.choice(statuses),
            "source_file": "orders.csv"
        })

    # Convert order records into a DataFrame.
    orders_df = pd.DataFrame(orders)

    # Create a separate list of expected rejection labels.
    expected_rejects = []

    if inject_issues and order_count >= 8:

        # Scenario 1: Duplicate order ID.
        orders_df.loc[1, "order_id"] = orders_df.loc[0, "order_id"]

        expected_rejects.append({
            "source_record_id": orders_df.loc[0, "source_record_id"],
            "expected_reason": "duplicate_order_id"
        })

        expected_rejects.append({
            "source_record_id": orders_df.loc[1, "source_record_id"],
            "expected_reason": "duplicate_order_id"
        })

        # Scenario 2: Missing customer ID.
        orders_df.loc[2, "customer_id"] = None

        expected_rejects.append({
            "source_record_id": orders_df.loc[2, "source_record_id"],
            "expected_reason": "missing_customer_id"
        })

        # Scenario 3: Customer reference does not exist.
        orders_df.loc[3, "customer_id"] = "C999999"

        expected_rejects.append({
            "source_record_id": orders_df.loc[3, "source_record_id"],
            "expected_reason": "invalid_customer_reference"
        })

        # Scenario 4: Invalid quantity.
        orders_df.loc[4, "quantity"] = 0

        expected_rejects.append({
            "source_record_id": orders_df.loc[4, "source_record_id"],
            "expected_reason": "invalid_quantity"
        })

        # Scenario 5: Negative price.
        orders_df.loc[5, "unit_price"] = -50

        expected_rejects.append({
            "source_record_id": orders_df.loc[5, "source_record_id"],
            "expected_reason": "negative_unit_price"
        })

        # Scenario 6: Unsupported status.
        orders_df.loc[6, "status"] = "UNKNOWN"

        expected_rejects.append({
            "source_record_id": orders_df.loc[6, "source_record_id"],
            "expected_reason": "invalid_status"
        })

        # Scenario 7: Malformed timestamp.
        orders_df.loc[7, "order_timestamp"] = "NOT_A_TIMESTAMP"

        expected_rejects.append({
            "source_record_id": orders_df.loc[7, "source_record_id"],
            "expected_reason": "invalid_timestamp"
        })

    # Convert expected rejection labels into a DataFrame.
    expected_rejects_df = pd.DataFrame(expected_rejects)

    return customers_df, orders_df, expected_rejects_df


def main():

    # Read generator settings from the command line.
    parser = argparse.ArgumentParser(
        description="Generate synthetic e-commerce data."
    )

    parser.add_argument("--customers", type=int, default=100)
    parser.add_argument("--orders", type=int, default=200)
    parser.add_argument("--seed", type=int, default=42)
    parser.add_argument("--inject-issues", action="store_true")

    args = parser.parse_args()

    # Define project output directories.
    project_root = Path(__file__).resolve().parent.parent

    raw_folder = project_root / "data" / "raw"
    sample_folder = project_root / "data" / "sample"

    # Create folders if they do not exist.
    raw_folder.mkdir(parents=True, exist_ok=True)
    sample_folder.mkdir(parents=True, exist_ok=True)

    # Generate datasets.
    customers_df, orders_df, expected_rejects_df = generate_data(
        customer_count=args.customers,
        order_count=args.orders,
        seed=args.seed,
        inject_issues=args.inject_issues
    )

    # Save the source datasets.
    customers_df.to_csv(
        raw_folder / "customers.csv",
        index=False
    )

    orders_df.to_csv(
        raw_folder / "orders.csv",
        index=False
    )

    # Save expected rejection labels separately.
    expected_rejects_df.to_csv(
        sample_folder / "expected_rejects.csv",
        index=False
    )

    # Display generation summary.
    print("DATA GENERATION COMPLETED")
    print("-------------------------")
    print("Customers generated:", len(customers_df))
    print("Orders generated:", len(orders_df))
    print("Expected rejection labels:", len(expected_rejects_df))
    print("Random seed:", args.seed)
    print("Raw data folder:", raw_folder)
    print("Expected labels folder:", sample_folder)


if __name__ == "__main__":
    main()
```

```python

"""
Stage 2: Data validation module

Responsibilities:
1. Validate customer records.
2. Validate order records.
3. Collect all rejection reasons.
4. Separate valid and rejected records.
5. Save validation outputs and summary.
"""

from pathlib import Path
import pandas as pd
import json


# ---------------------------------------------------------
# PROJECT PATHS AND BUSINESS RULES
# ---------------------------------------------------------

PROJECT_ROOT = Path.cwd()
RAW_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
QUARANTINE_DIR = PROJECT_ROOT / "data" / "quarantine"

ALLOWED_STATUSES = {
    "PLACED",
    "SHIPPED",
    "DELIVERED",
    "CANCELLED"
}


# ---------------------------------------------------------
# HELPER FUNCTIONS
# ---------------------------------------------------------

def is_blank(series):
    """Identify missing or whitespace-only values."""
    return series.isna() | series.astype("string").str.strip().eq("")


def duplicate_mask(series):
    """Mark every occurrence of a duplicated nonblank value."""
    values = series.astype("string").str.strip()
    return (
        values.notna()
        & values.ne("")
        & values.duplicated(keep=False)
    )


def add_reason(reasons, mask, reason):
    """Add a rejection reason to every row matching a condition."""
    for position, failed in enumerate(mask.fillna(False).tolist()):
        if failed:
            reasons[position].append(reason)


def separate_records(df, reasons):
    """Separate valid and rejected records."""
    result = df.copy()

    # Store every rejection reason for each rejected record.
    result["rejection_reasons"] = [
        ";".join(row_reasons) for row_reasons in reasons
    ]

    rejected_mask = result["rejection_reasons"].ne("")

    valid_records = result.loc[~rejected_mask].drop(
        columns=["rejection_reasons"]
    ).copy()

    rejected_records = result.loc[rejected_mask].copy()

    return valid_records, rejected_records


# ---------------------------------------------------------
# CUSTOMER VALIDATION
# ---------------------------------------------------------

def validate_customers(customers):
    """Validate customer records and return valid/rejected data."""

    df = customers.copy().reset_index(drop=True)
    reasons = [[] for _ in range(len(df))]

    # Required fields.
    required_fields = [
        "customer_id",
        "customer_name",
        "email",
        "created_at"
    ]

    for column in required_fields:
        add_reason(
            reasons,
            is_blank(df[column]),
            f"missing_{column}"
        )

    # Customer IDs must be unique.
    add_reason(
        reasons,
        duplicate_mask(df["customer_id"]),
        "duplicate_customer_id"
    )

    # Basic email format validation.
    email = df["email"].astype("string").str.strip()

    valid_email = email.str.match(
        r"^[^@\s]+@[^@\s]+\.[^@\s]+$",
        na=False
    )

    add_reason(
        reasons,
        ~is_blank(df["email"]) & ~valid_email,
        "invalid_email"
    )

    # Validate customer timestamps.
    parsed_dates = pd.to_datetime(
        df["created_at"],
        errors="coerce",
        utc=True
    )

    add_reason(
        reasons,
        ~is_blank(df["created_at"]) & parsed_dates.isna(),
        "invalid_created_at"
    )

    return separate_records(df, reasons)


# ---------------------------------------------------------
# ORDER VALIDATION
# ---------------------------------------------------------

def validate_orders(orders, valid_customer_ids):
    """Validate order records against business rules."""

    df = orders.copy().reset_index(drop=True)
    reasons = [[] for _ in range(len(df))]

    # Required fields.
    required_fields = [
        "source_record_id",
        "order_id",
        "customer_id",
        "order_timestamp",
        "quantity",
        "unit_price",
        "status",
        "source_file"
    ]

    for column in required_fields:
        add_reason(
            reasons,
            is_blank(df[column]),
            f"missing_{column}"
        )

    # Both source record IDs and order IDs must be unique.
    add_reason(
        reasons,
        duplicate_mask(df["source_record_id"]),
        "duplicate_source_record_id"
    )

    add_reason(
        reasons,
        duplicate_mask(df["order_id"]),
        "duplicate_order_id"
    )

    # Customer must exist in the validated customer dataset.
    customer_id = df["customer_id"].astype("string").str.strip()

    add_reason(
        reasons,
        ~is_blank(df["customer_id"])
        & ~customer_id.isin(valid_customer_ids),
        "customer_not_found"
    )

    # Validate timestamps.
    parsed_dates = pd.to_datetime(
        df["order_timestamp"],
        errors="coerce",
        utc=True
    )

    add_reason(
        reasons,
        ~is_blank(df["order_timestamp"]) & parsed_dates.isna(),
        "invalid_order_timestamp"
    )

    # Validate quantity.
    quantity = pd.to_numeric(
        df["quantity"],
        errors="coerce"
    )

    add_reason(
        reasons,
        ~is_blank(df["quantity"]) & quantity.isna(),
        "quantity_must_be_numeric"
    )

    add_reason(
        reasons,
        quantity.notna() & (quantity <= 0),
        "quantity_must_be_positive"
    )

    add_reason(
        reasons,
        quantity.notna() & (quantity % 1 != 0),
        "quantity_must_be_integer"
    )

    # Validate unit price.
    price = pd.to_numeric(
        df["unit_price"],
        errors="coerce"
    )

    add_reason(
        reasons,
        ~is_blank(df["unit_price"]) & price.isna(),
        "unit_price_must_be_numeric"
    )

    add_reason(
        reasons,
        price.notna() & (price < 0),
        "unit_price_must_be_non_negative"
    )

    # Validate allowed order statuses.
    status = df["status"].astype("string").str.strip().str.upper()

    add_reason(
        reasons,
        ~is_blank(df["status"])
        & ~status.isin(ALLOWED_STATUSES),
        "invalid_order_status"
    )

    return separate_records(df, reasons)


# ---------------------------------------------------------
# SAVE VALIDATION OUTPUTS
# ---------------------------------------------------------

def save_outputs(
    valid_customers,
    rejected_customers,
    valid_orders,
    rejected_orders
):
    """Save accepted and rejected records into separate CSV files."""

    PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
    QUARANTINE_DIR.mkdir(parents=True, exist_ok=True)

    valid_customers.to_csv(
        PROCESSED_DIR / "valid_customers.csv",
        index=False
    )

    valid_orders.to_csv(
        PROCESSED_DIR / "valid_orders.csv",
        index=False
    )

    rejected_customers.to_csv(
        QUARANTINE_DIR / "rejected_customers.csv",
        index=False
    )

    rejected_orders.to_csv(
        QUARANTINE_DIR / "rejected_orders.csv",
        index=False
    )


# ---------------------------------------------------------
# MAIN VALIDATION WORKFLOW
# ---------------------------------------------------------

def run_validation():
    """Execute the complete Stage 2 validation workflow."""

    # Read the original source files.
    customers = pd.read_csv(RAW_DIR / "customers.csv")
    orders = pd.read_csv(RAW_DIR / "orders.csv")

    # Validate customers first.
    valid_customers, rejected_customers = validate_customers(
        customers
    )

    # Only successfully validated customer IDs are accepted.
    valid_customer_ids = set(
        valid_customers["customer_id"].astype(str).str.strip()
    )

    # Validate orders using the valid customer IDs.
    valid_orders, rejected_orders = validate_orders(
        orders,
        valid_customer_ids
    )

    # Save all four output files.
    save_outputs(
        valid_customers,
        rejected_customers,
        valid_orders,
        rejected_orders
    )

    # Create a summary.
    summary = {
        "customers": {
            "input": len(customers),
            "accepted": len(valid_customers),
            "rejected": len(rejected_customers)
        },
        "orders": {
            "input": len(orders),
            "accepted": len(valid_orders),
            "rejected": len(rejected_orders)
        }
    }

    # Save summary as JSON.
    report_path = PROJECT_ROOT / "reports"
    report_path.mkdir(parents=True, exist_ok=True)

    with open(
        report_path / "validation_summary.json",
        "w",
        encoding="utf-8"
    ) as file:
        json.dump(summary, file, indent=4)

    print("STAGE 2 VALIDATION COMPLETED")
    print("--------------------------------")

    print("Customers:", summary["customers"])
    print("Orders:", summary["orders"])

    print("--------------------------------")
    print("Validation files saved successfully.")

    return summary
```

```python
%%writefile src/validation.py

"""
Stage 2: Data validation module.

Validates customer and order records, separates accepted and
rejected records, and generates a validation summary.
"""

from pathlib import Path
import pandas as pd
import json


# Project directories.
PROJECT_ROOT = Path(__file__).resolve().parent.parent

RAW_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
QUARANTINE_DIR = PROJECT_ROOT / "data" / "quarantine"
REPORT_DIR = PROJECT_ROOT / "reports"

# Allowed order statuses.
ALLOWED_STATUSES = {
    "PLACED", "SHIPPED", "DELIVERED", "CANCELLED"
}


# Identify missing or blank values.
def is_blank(series):
    return (
        series.isna()
        | series.astype("string").str.strip().eq("")
    )


# Mark every occurrence of a duplicate value.
def duplicate_mask(series):
    values = series.astype("string").str.strip()

    return (
        values.notna()
        & values.ne("")
        & values.duplicated(keep=False)
    )


# Attach all applicable rejection reasons to each row.
def add_reason(reasons, mask, reason):
    for position, failed in enumerate(mask.fillna(False).tolist()):
        if failed:
            reasons[position].append(reason)


# Separate accepted and rejected records.
def separate_records(df, reasons):
    result = df.copy()

    result["rejection_reasons"] = [
        ";".join(row_reasons)
        for row_reasons in reasons
    ]

    rejected_mask = result["rejection_reasons"].ne("")

    valid = result.loc[~rejected_mask].copy()

    rejected = result.loc[rejected_mask].copy()

    valid = valid.drop(columns=["rejection_reasons"])

    return valid, rejected


# --------------------------------------------------
# CUSTOMER VALIDATION
# --------------------------------------------------

def validate_customers(customers):

    df = customers.copy().reset_index(drop=True)

    reasons = [[] for _ in range(len(df))]

    required_fields = [
        "customer_id",
        "customer_name",
        "email",
        "created_at"
    ]

    for column in required_fields:
        add_reason(
            reasons,
            is_blank(df[column]),
            f"missing_{column}"
        )

    # Check duplicate customer IDs.
    add_reason(
        reasons,
        duplicate_mask(df["customer_id"]),
        "duplicate_customer_id"
    )

    # Check basic email format.
    email = df["email"].astype("string").str.strip()

    valid_email = email.str.match(
        r"^[^@\s]+@[^@\s]+\.[^@\s]+$",
        na=False
    )

    add_reason(
        reasons,
        ~is_blank(df["email"]) & ~valid_email,
        "invalid_email"
    )

    # Check customer timestamps.
    parsed_dates = pd.to_datetime(
        df["created_at"],
        errors="coerce",
        utc=True
    )

    add_reason(
        reasons,
        ~is_blank(df["created_at"]) & parsed_dates.isna(),
        "invalid_created_at"
    )

    return separate_records(df, reasons)


# --------------------------------------------------
# ORDER VALIDATION
# --------------------------------------------------

def validate_orders(orders, valid_customer_ids):

    df = orders.copy().reset_index(drop=True)

    reasons = [[] for _ in range(len(df))]

    required_fields = [
        "source_record_id",
        "order_id",
        "customer_id",
        "order_timestamp",
        "quantity",
        "unit_price",
        "status",
        "source_file"
    ]

    for column in required_fields:
        add_reason(
            reasons,
            is_blank(df[column]),
            f"missing_{column}"
        )

    # Check duplicate source record IDs.
    add_reason(
        reasons,
        duplicate_mask(df["source_record_id"]),
        "duplicate_source_record_id"
    )

    # Reject every occurrence of duplicate order IDs.
    add_reason(
        reasons,
        duplicate_mask(df["order_id"]),
        "duplicate_order_id"
    )

    # Check whether customer references exist.
    customer_ids = (
        df["customer_id"]
        .astype("string")
        .str.strip()
    )

    add_reason(
        reasons,
        ~is_blank(df["customer_id"])
        & ~customer_ids.isin(valid_customer_ids),
        "customer_not_found"
    )

    # Check order timestamps.
    parsed_dates = pd.to_datetime(
        df["order_timestamp"],
        errors="coerce",
        utc=True
    )

    add_reason(
        reasons,
        ~is_blank(df["order_timestamp"])
        & parsed_dates.isna(),
        "invalid_order_timestamp"
    )

    # Check quantity values.
    quantity = pd.to_numeric(
        df["quantity"],
        errors="coerce"
    )

    add_reason(
        reasons,
        ~is_blank(df["quantity"]) & quantity.isna(),
        "quantity_must_be_numeric"
    )

    add_reason(
        reasons,
        quantity.notna() & (quantity <= 0),
        "quantity_must_be_positive"
    )

    add_reason(
        reasons,
        quantity.notna() & (quantity % 1 != 0),
        "quantity_must_be_integer"
    )

    # Check unit prices.
    price = pd.to_numeric(
        df["unit_price"],
        errors="coerce"
    )

    add_reason(
        reasons,
        ~is_blank(df["unit_price"]) & price.isna(),
        "unit_price_must_be_numeric"
    )

    add_reason(
        reasons,
        price.notna() & (price < 0),
        "unit_price_must_be_non_negative"
    )

    # Check allowed statuses.
    status = (
        df["status"]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    add_reason(
        reasons,
        ~is_blank(df["status"])
        & ~status.isin(ALLOWED_STATUSES),
        "invalid_order_status"
    )

    return separate_records(df, reasons)


# --------------------------------------------------
# SAVE OUTPUT FILES
# --------------------------------------------------

def save_outputs(
    valid_customers,
    rejected_customers,
    valid_orders,
    rejected_orders
):

    PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
    QUARANTINE_DIR.mkdir(parents=True, exist_ok=True)

    valid_customers.to_csv(
        PROCESSED_DIR / "valid_customers.csv",
        index=False
    )

    valid_orders.to_csv(
        PROCESSED_DIR / "valid_orders.csv",
        index=False
    )

    rejected_customers.to_csv(
        QUARANTINE_DIR / "rejected_customers.csv",
        index=False
    )

    rejected_orders.to_csv(
        QUARANTINE_DIR / "rejected_orders.csv",
        index=False
    )


# --------------------------------------------------
# MAIN VALIDATION WORKFLOW
# --------------------------------------------------

def run_validation():

    # Read source files.
    customers = pd.read_csv(RAW_DIR / "customers.csv")

    orders = pd.read_csv(RAW_DIR / "orders.csv")

    # Validate customers first.
    valid_customers, rejected_customers = (
        validate_customers(customers)
    )

    # Use only accepted customer IDs for order validation.
    valid_customer_ids = set(
        valid_customers["customer_id"]
        .astype(str)
        .str.strip()
    )

    # Validate orders.
    valid_orders, rejected_orders = validate_orders(
        orders,
        valid_customer_ids
    )

    # Save all accepted and rejected files.
    save_outputs(
        valid_customers,
        rejected_customers,
        valid_orders,
        rejected_orders
    )

    # Build validation summary.
    summary = {
        "customers": {
            "input": len(customers),
            "accepted": len(valid_customers),
            "rejected": len(rejected_customers)
        },
        "orders": {
            "input": len(orders),
            "accepted": len(valid_orders),
            "rejected": len(rejected_orders)
        }
    }

    # Save summary as JSON.
    REPORT_DIR.mkdir(parents=True, exist_ok=True)

    with open(
        REPORT_DIR / "validation_summary.json",
        "w",
        encoding="utf-8"
    ) as file:
        json.dump(summary, file, indent=4)

    print("STAGE 2 VALIDATION COMPLETED")
    print("----------------------------------")
    print("Customers:", summary["customers"])
    print("Orders:", summary["orders"])
    print("----------------------------------")

    return summary
```

```python

%%writefile src/transformation.py

from pathlib import Path
import json
import pandas as pd


# Find the main project folder.
PROJECT_ROOT = Path(__file__).resolve().parent.parent

# Define input and output folders.
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
REPORT_DIR = PROJECT_ROOT / "reports"


def transform_customers(customers):
    """
    Standardise customer data.

    Input:
        customers: DataFrame containing validated customers.

    Output:
        A transformed customer DataFrame.
    """

    customers = customers.copy()

    # Remove unnecessary spaces from text columns.
    for column in ["customer_id", "customer_name", "email", "city"]:
        if column in customers.columns:
            customers[column] = customers[column].astype("string").str.strip()

    # Convert customer creation timestamps to UTC.
    customers["created_at"] = pd.to_datetime(
        customers["created_at"],
        errors="raise",
        utc=True
    )

    # Store timestamps in ISO-8601 format.
    customers["created_at"] = customers["created_at"].apply(
        lambda value: value.isoformat()
    )

    return customers


def transform_orders(orders):
    """
    Standardise orders and calculate total_amount.

    Input:
        orders: DataFrame containing validated orders.

    Output:
        A transformed order DataFrame.
    """

    orders = orders.copy()

    # Remove unnecessary spaces from text columns.
    for column in [
        "source_record_id",
        "order_id",
        "customer_id",
        "status",
        "source_file"
    ]:
        orders[column] = orders[column].astype("string").str.strip()

    # Standardise order status to uppercase.
    orders["status"] = orders["status"].str.upper()

    # Convert order timestamps to UTC.
    orders["order_timestamp"] = pd.to_datetime(
        orders["order_timestamp"],
        errors="raise",
        utc=True
    )

    # Store timestamps in ISO-8601 format.
    orders["order_timestamp"] = orders["order_timestamp"].apply(
        lambda value: value.isoformat()
    )

    # Convert quantity and unit price to numeric values.
    orders["quantity"] = pd.to_numeric(
        orders["quantity"],
        errors="raise"
    ).astype("int64")

    orders["unit_price"] = pd.to_numeric(
        orders["unit_price"],
        errors="raise"
    )

    # Calculate the total order amount.
    # Round to two decimal places for monetary output.
    orders["total_amount"] = (
        orders["quantity"] * orders["unit_price"]
    ).round(2)

    return orders


def run_transformation():

    # Read validated input files.
    customers_path = PROCESSED_DIR / "valid_customers.csv"
    orders_path = PROCESSED_DIR / "valid_orders.csv"

    customers = pd.read_csv(customers_path)
    orders = pd.read_csv(orders_path)

    # Transform both datasets.
    transformed_customers = transform_customers(customers)
    transformed_orders = transform_orders(orders)

    # Verify that transformation did not change row counts.
    assert len(transformed_customers) == len(customers)
    assert len(transformed_orders) == len(orders)

    # Verify that every order has a calculated amount.
    assert transformed_orders["total_amount"].notna().all()

    # Verify the calculated amounts are not negative.
    assert (transformed_orders["total_amount"] >= 0).all()

    # Save transformed datasets.
    customer_output = PROCESSED_DIR / "transformed_customers.csv"
    order_output = PROCESSED_DIR / "transformed_orders.csv"

    transformed_customers.to_csv(customer_output, index=False)
    transformed_orders.to_csv(order_output, index=False)

    # Prepare a transformation report.
    summary = {
        "customers": {
            "input": len(customers),
            "output": len(transformed_customers)
        },
        "orders": {
            "input": len(orders),
            "output": len(transformed_orders)
        },
        "total_order_amount": round(
            float(transformed_orders["total_amount"].sum()), 2
        ),
        "output_files": [
            str(customer_output.relative_to(PROJECT_ROOT)),
            str(order_output.relative_to(PROJECT_ROOT))
        ],
        "status": "SUCCESS"
    }

    # Save the report as JSON.
    REPORT_DIR.mkdir(parents=True, exist_ok=True)

    report_path = REPORT_DIR / "transformation_summary.json"

    with open(report_path, "w", encoding="utf-8") as file:
        json.dump(summary, file, indent=4)

    print("STAGE 3 TRANSFORMATION COMPLETED")
    print("----------------------------------")
    print("Customers:", summary["customers"])
    print("Orders:", summary["orders"])
    print("Total order amount:", summary["total_order_amount"])
    print("Status:", summary["status"])
    print("----------------------------------")

    return summary
```

```python

%%writefile src/publishing.py

from pathlib import Path
import sqlite3
import json
import pandas as pd
from datetime import datetime, timezone


# Find the main project directory.
PROJECT_ROOT = Path(__file__).resolve().parent.parent

# Define input, database and report locations.
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
WAREHOUSE_DIR = PROJECT_ROOT / "data" / "warehouse"
REPORT_DIR = PROJECT_ROOT / "reports"

# SQLite database file.
DATABASE_PATH = WAREHOUSE_DIR / "pipeline.db"


def create_tables(connection):
    """
    Create database tables if they do not already exist.

    Primary keys prevent duplicate business records.
    Foreign keys maintain relationships between customers and orders.
    """

    cursor = connection.cursor()

    # Create the customers table.
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS customers (
            customer_id TEXT PRIMARY KEY,
            customer_name TEXT NOT NULL,
            email TEXT NOT NULL,
            city TEXT,
            created_at TEXT NOT NULL
        )
    """)

    # Create the orders table.
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS orders (
            order_id TEXT PRIMARY KEY,
            source_record_id TEXT NOT NULL UNIQUE,
            customer_id TEXT NOT NULL,
            order_timestamp TEXT NOT NULL,
            quantity INTEGER NOT NULL CHECK(quantity > 0),
            unit_price REAL NOT NULL CHECK(unit_price >= 0),
            total_amount REAL NOT NULL CHECK(total_amount >= 0),
            status TEXT NOT NULL CHECK(
                status IN ('PLACED', 'SHIPPED', 'DELIVERED', 'CANCELLED')
            ),
            source_file TEXT,
            FOREIGN KEY(customer_id) REFERENCES customers(customer_id)
        )
    """)

    connection.commit()


def publish_customers(connection, customers):
    """
    Insert new customers or update existing customers.

    ON CONFLICT makes repeated pipeline runs safe for the same
    customer IDs.
    """

    records = customers[
        [
            "customer_id",
            "customer_name",
            "email",
            "city",
            "created_at"
        ]
    ].where(pd.notna(customers), None)

    rows = list(records.itertuples(index=False, name=None))

    connection.executemany("""
        INSERT INTO customers (
            customer_id,
            customer_name,
            email,
            city,
            created_at
        )
        VALUES (?, ?, ?, ?, ?)

        ON CONFLICT(customer_id) DO UPDATE SET
            customer_name = excluded.customer_name,
            email = excluded.email,
            city = excluded.city,
            created_at = excluded.created_at
    """, rows)

    return len(rows)


def publish_orders(connection, orders):
    """
    Insert new orders or update existing orders.

    ON CONFLICT(order_id) prevents duplicate orders during reruns.
    """

    records = orders[
        [
            "order_id",
            "source_record_id",
            "customer_id",
            "order_timestamp",
            "quantity",
            "unit_price",
            "total_amount",
            "status",
            "source_file"
        ]
    ].where(pd.notna(orders), None)

    rows = list(records.itertuples(index=False, name=None))

    connection.executemany("""
        INSERT INTO orders (
            order_id,
            source_record_id,
            customer_id,
            order_timestamp,
            quantity,
            unit_price,
            total_amount,
            status,
            source_file
        )
        VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?)

        ON CONFLICT(order_id) DO UPDATE SET
            source_record_id = excluded.source_record_id,
            customer_id = excluded.customer_id,
            order_timestamp = excluded.order_timestamp,
            quantity = excluded.quantity,
            unit_price = excluded.unit_price,
            total_amount = excluded.total_amount,
            status = excluded.status,
            source_file = excluded.source_file
    """, rows)

    return len(rows)


def run_publishing():

    # Read transformed CSV files.
    customers = pd.read_csv(
        PROCESSED_DIR / "transformed_customers.csv"
    )

    orders = pd.read_csv(
        PROCESSED_DIR / "transformed_orders.csv"
    )

    # Create database directory if required.
    WAREHOUSE_DIR.mkdir(parents=True, exist_ok=True)

    # Open SQLite connection.
    connection = sqlite3.connect(DATABASE_PATH)

    # Enable foreign-key enforcement for this connection.
    connection.execute("PRAGMA foreign_keys = ON")

    try:
        # Create database tables.
        create_tables(connection)

        # Start one transaction for both publishing operations.
        connection.execute("BEGIN")

        # Publish customers first because orders reference customers.
        customer_count = publish_customers(connection, customers)

        # Publish orders after customers.
        order_count = publish_orders(connection, orders)

        # Reconcile database counts.
        database_customer_count = connection.execute(
            "SELECT COUNT(*) FROM customers"
        ).fetchone()[0]

        database_order_count = connection.execute(
            "SELECT COUNT(*) FROM orders"
        ).fetchone()[0]

        # Confirm all incoming records are represented.
        assert database_customer_count >= len(customers)
        assert database_order_count >= len(orders)

        # Calculate the total amount stored in the database.
        database_total_amount = connection.execute(
            "SELECT ROUND(SUM(total_amount), 2) FROM orders"
        ).fetchone()[0]

        # Commit both publishing operations together.
        connection.commit()

        summary = {
            "status": "SUCCESS",
            "database": str(
                DATABASE_PATH.relative_to(PROJECT_ROOT)
            ),
            "customers_processed": customer_count,
            "orders_processed": order_count,
            "customers_in_database": database_customer_count,
            "orders_in_database": database_order_count,
            "database_total_amount": database_total_amount,
            "published_at": datetime.now(
                timezone.utc
            ).isoformat()
        }

    except Exception:
        # Roll back database changes if publishing fails.
        connection.rollback()
        raise

    finally:
        # Always close the connection.
        connection.close()

    # Save the publishing report.
    REPORT_DIR.mkdir(parents=True, exist_ok=True)

    report_path = REPORT_DIR / "publishing_summary.json"

    with open(report_path, "w", encoding="utf-8") as file:
        json.dump(summary, file, indent=4)

    print("STAGE 4 PUBLISHING COMPLETED")
    print("----------------------------------")
    print("Customers processed:", summary["customers_processed"])
    print("Orders processed:", summary["orders_processed"])
    print("Customers in database:", summary["customers_in_database"])
    print("Orders in database:", summary["orders_in_database"])
    print("Database total amount:", summary["database_total_amount"])
    print("Status:", summary["status"])
    print("----------------------------------")

    return summary
```

```python

%%writefile src/orchestrator.py

from pathlib import Path
from datetime import datetime, timezone
import subprocess
import sys
import json
import time
import uuid
import logging
import os


# Locate the main project folder.
PROJECT_ROOT = Path(__file__).resolve().parent.parent

# Create folders for logs and reports.
LOG_DIR = PROJECT_ROOT / "logs"
REPORT_DIR = PROJECT_ROOT / "reports"

LOG_DIR.mkdir(parents=True, exist_ok=True)
REPORT_DIR.mkdir(parents=True, exist_ok=True)

# Configure pipeline logging.
LOG_FILE = LOG_DIR / "pipeline.log"

logging.basicConfig(
    filename=LOG_FILE,
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s",
    force=True
)

logger = logging.getLogger("pipeline")


# Define the tasks and their execution order.
# Each task runs in a separate Python process.
TASKS = [
    {
        "name": "validation",
        "description": "Read raw files and validate customer and order data",
        "code": "from src.validation import run_validation; run_validation()",
        "max_retries": 3,
        "timeout_seconds": 120
    },
    {
        "name": "transformation",
        "description": "Transform validated customer and order data",
        "code": "from src.transformation import run_transformation; run_transformation()",
        "max_retries": 3,
        "timeout_seconds": 120
    },
    {
        "name": "publishing",
        "description": "Publish transformed records into SQLite",
        "code": "from src.publishing import run_publishing; run_publishing()",
        "max_retries": 3,
        "timeout_seconds": 120
    }
]


def save_json(path, data):
    """Save a Python dictionary as a formatted JSON file."""

    with open(path, "w", encoding="utf-8") as file:
        json.dump(data, file, indent=4)


def run_task(task, run_id):
    """
    Execute one pipeline task with retry and timeout handling.

    A task is successful only when its Python process exits
    with return code zero.
    """

    task_name = task["name"]
    max_attempts = task["max_retries"]
    timeout_seconds = task["timeout_seconds"]

    for attempt in range(1, max_attempts + 1):

        start_time = time.monotonic()

        logger.info(
            "RUN %s | TASK %s | ATTEMPT %s STARTED",
            run_id,
            task_name,
            attempt
        )

        print(
            f"\n[{task_name.upper()}] "
            f"Attempt {attempt}/{max_attempts}"
        )

        # Pass the run ID to the child Python process.
        environment = os.environ.copy()
        environment["PIPELINE_RUN_ID"] = run_id

        try:

            # Execute the task in a separate Python process.
            result = subprocess.run(
                [
                    sys.executable,
                    "-c",
                    task["code"]
                ],
                cwd=PROJECT_ROOT,
                capture_output=True,
                text=True,
                timeout=timeout_seconds,
                env=environment
            )

            duration = round(
                time.monotonic() - start_time, 2
            )

            if result.stdout:
                print(result.stdout)

            if result.returncode == 0:

                logger.info(
                    "RUN %s | TASK %s | SUCCESS | ATTEMPT %s",
                    run_id,
                    task_name,
                    attempt
                )

                return {
                    "task": task_name,
                    "status": "SUCCESS",
                    "attempts": attempt,
                    "duration_seconds": duration,
                    "error": None
                }

            # A non-zero exit code means the task failed.
            error_message = (
                result.stderr.strip()
                or f"Process exited with code {result.returncode}"
            )

        except subprocess.TimeoutExpired:

            duration = round(
                time.monotonic() - start_time, 2
            )

            error_message = (
                f"Task exceeded {timeout_seconds} seconds"
            )

        except Exception as error:

            duration = round(
                time.monotonic() - start_time, 2
            )

            error_message = str(error)

        logger.error(
            "RUN %s | TASK %s | ATTEMPT %s FAILED | %s",
            run_id,
            task_name,
            attempt,
            error_message
        )

        print("Task failed:", error_message)

        # Retry if attempts are still available.
        if attempt < max_attempts:
            print("Retrying task...")
            time.sleep(2)

    # Return failure after all attempts are exhausted.
    return {
        "task": task_name,
        "status": "FAILED",
        "attempts": max_attempts,
        "duration_seconds": duration,
        "error": error_message
    }


def run_pipeline():

    # Generate a unique ID for this pipeline execution.
    run_id = str(uuid.uuid4())

    pipeline_start = datetime.now(timezone.utc)
    start_time = time.monotonic()

    print("\n======================================")
    print("DATA ORCHESTRATION PIPELINE STARTED")
    print("Run ID:", run_id)
    print("Started at:", pipeline_start.isoformat())
    print("======================================")

    logger.info("PIPELINE %s STARTED", run_id)

    task_results = []
    pipeline_failed = False

    # Execute tasks in dependency order.
    for index, task in enumerate(TASKS):

        if pipeline_failed:

            # Never execute downstream tasks after a failure.
            task_results.append({
                "task": task["name"],
                "status": "SKIPPED",
                "attempts": 0,
                "duration_seconds": 0,
                "error": "An upstream task failed"
            })

            print(
                f"\n[{task['name'].upper()}] SKIPPED "
                "- upstream failure"
            )

            continue

        result = run_task(task, run_id)
        task_results.append(result)

        if result["status"] == "FAILED":
            pipeline_failed = True

    # Calculate total execution duration.
    total_duration = round(
        time.monotonic() - start_time, 2
    )

    pipeline_end = datetime.now(timezone.utc)

    # Determine overall pipeline status.
    overall_status = (
        "FAILED" if pipeline_failed else "SUCCESS"
    )

    # Build the final execution report.
    summary = {
        "run_id": run_id,
        "status": overall_status,
        "started_at": pipeline_start.isoformat(),
        "finished_at": pipeline_end.isoformat(),
        "duration_seconds": total_duration,
        "tasks": task_results
    }

    # Save the latest run summary.
    summary_path = REPORT_DIR / "pipeline_run_summary.json"
    save_json(summary_path, summary)

    # Preserve the history of all pipeline runs.
    history_path = REPORT_DIR / "run_history.jsonl"

    with open(history_path, "a", encoding="utf-8") as file:
        file.write(json.dumps(summary) + "\n")

    logger.info(
        "PIPELINE %s FINISHED | STATUS %s",
        run_id,
        overall_status
    )

    print("\n======================================")
    print("PIPELINE EXECUTION SUMMARY")
    print("======================================")

    for result in task_results:
        print(
            result["task"].upper(),
            "|",
            result["status"],
            "| Attempts:",
            result["attempts"]
        )

    print("--------------------------------------")
    print("Overall status:", overall_status)
    print("Total duration:", total_duration, "seconds")
    print("Run ID:", run_id)
    print("======================================")

    if pipeline_failed:
        print("ALERT: Pipeline failed. Check the logs and reports.")

    else:
        print("SUCCESS: All pipeline tasks completed.")

    return summary


if __name__ == "__main__":
    run_pipeline()
```

```python

%%writefile tests/test_orchestrator.py

from pathlib import Path
from unittest.mock import patch

from src import orchestrator


def test_run_task_success():
    task = {
        "name": "validation",
        "code": "print('Task completed')",
        "max_retries": 3,
        "timeout_seconds": 10
    }

    with patch("src.orchestrator.subprocess.run") as mock_run:
        mock_run.return_value.returncode = 0
        mock_run.return_value.stdout = "Task completed"
        mock_run.return_value.stderr = ""

        result = orchestrator.run_task(task, "TEST_RUN_001")

    assert result["status"] == "SUCCESS"
    assert result["attempts"] == 1
    assert result["error"] is None


def test_run_task_retries_after_failure():
    task = {
        "name": "validation",
        "code": "print('Testing')",
        "max_retries": 2,
        "timeout_seconds": 10
    }

    with patch("src.orchestrator.subprocess.run") as mock_run, \
         patch("src.orchestrator.time.sleep"):

        mock_run.return_value.returncode = 1
        mock_run.return_value.stdout = ""
        mock_run.return_value.stderr = "Simulated failure"

        result = orchestrator.run_task(task, "TEST_RUN_002")

    assert mock_run.call_count == 2
    assert result["status"] == "FAILED"
    assert result["attempts"] == 2
    assert result["error"] == "Simulated failure"


def test_run_task_handles_timeout():
    task = {
        "name": "validation",
        "code": "print('Testing')",
        "max_retries": 1,
        "timeout_seconds": 1
    }

    with patch(
        "src.orchestrator.subprocess.run",
        side_effect=orchestrator.subprocess.TimeoutExpired(
            cmd="test",
            timeout=1
        )
    ):

        result = orchestrator.run_task(task, "TEST_RUN_003")

    assert result["status"] == "FAILED"
    assert result["attempts"] == 1
    assert "exceeded 1 seconds" in result["error"]


def test_pipeline_skips_downstream_after_failure():
    test_tasks = [
        {
            "name": "validation",
            "code": "",
            "max_retries": 2,
            "timeout_seconds": 10
        },
        {
            "name": "transformation",
            "code": "",
            "max_retries": 2,
            "timeout_seconds": 10
        },
        {
            "name": "publishing",
            "code": "",
            "max_retries": 2,
            "timeout_seconds": 10
        }
    ]

    # Create the temporary report folder before saving reports.
    temporary_report_dir = Path("tests/temp_reports")
    temporary_report_dir.mkdir(parents=True, exist_ok=True)

    with patch.object(orchestrator, "TASKS", test_tasks):

        with patch.object(
            orchestrator,
            "run_task",
            return_value={
                "task": "validation",
                "status": "FAILED",
                "attempts": 2,
                "duration_seconds": 0.1,
                "error": "Simulated failure"
            }
        ) as mock_run_task:

            with patch.object(
                orchestrator,
                "REPORT_DIR",
                temporary_report_dir
            ):

                result = orchestrator.run_pipeline()

    # Only validation should have executed.
    assert mock_run_task.call_count == 1

    # Check the overall pipeline status.
    assert result["status"] == "FAILED"

    # Check that downstream tasks were skipped.
    assert result["tasks"][1]["status"] == "SKIPPED"
    assert result["tasks"][2]["status"] == "SKIPPED"
```

```python

import subprocess
import sys
import time

# Run the existing generator using the current Python environment.
command = [
    sys.executable,
    "src/generate_data.py",
    "--customers", "10000",
    "--orders", "100000",
    "--seed", "42"
]

# Measure how long data generation takes.
start_time = time.perf_counter()

result = subprocess.run(
    command,
    capture_output=True,
    text=True,
    cwd="."
)

end_time = time.perf_counter()

# Display the result.
print(result.stdout)

if result.returncode != 0:
    print("ERROR:")
    print(result.stderr)
else:
    print(f"Generation time: {end_time - start_time:.2f} seconds")
```
